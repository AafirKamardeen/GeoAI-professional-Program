```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
```

```python
landsat=rasterio.open("landsat.tif")
```

```python
print("Width:",landsat.width)
print("Height:",landsat.height)
print("Number of bands:",landsat.count)
```

```python
print("CRS", landsat.crs)
```

```python
print(
    "pixel area:",
    landsat.res[0]*landsat.res[1],
    "m2"
)
```

```python
print(landsat.bounds)
```

```python
print(landsat.transform)
```

```python
print(landsat.meta)
```

```python
band1=landsat.read(1, masked=True)
```

```python
plt.figure(figsize=(8,6))

plt.imshow(band1)

plt.colorbar(label="pixel value")

plt.title("chandigarah landsat-band1")

plt.show()
```
