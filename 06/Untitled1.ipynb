```python
import matplotlib.pyplot as plt


df.groupby("region")["temperature"].mean().plot(kind="bar")

plt.xlabel("Region")
plt.ylabel("Average Temperature")
plt.title("Average Temperature by Region")
plt.show()
```

```python
import matplotlib.pyplot as plt


df.groupby("region")["rainfall"].mean().plot(kind="bar")

plt.xlabel("region")
plt.ylabel("Average Rainfall")
plt.title("Average Rainfall byRegion")
plt.show()
```

```python
import matplotlib.pyplot as plt


df.groupby("region")["ndvi"].mean().plot(kind="pie")

plt.xlabel("Region")
plt.ylabel("Average NDVI")
plt.title("Average NDVI by Region")
plt.show()
```
