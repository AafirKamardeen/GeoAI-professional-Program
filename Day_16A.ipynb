```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import rasterio
from rasterio.features import geometry_mask

# =========================================================
# 0. FILE PATHS
# =========================================================
GREEN_PATH = r"B3.tif"    # Sentinel-2 B3 (Green)
NIR_PATH   = r"B8.tif"    # Sentinel-2 B8 (NIR)
SHP_PATH   = r"chennai.shp"

# If chennai.shp is not in this folder, search the "Geo AI" folder for it
if not os.path.exists(SHP_PATH):
    search_root = os.path.dirname(os.getcwd())          # one level up (Geo AI)
    found = glob.glob(os.path.join(search_root, "**", "*hennai*.shp"), recursive=True)
    if found:
        SHP_PATH = found[0]
        print("Found shapefile at:", SHP_PATH)
    else:
        raise FileNotFoundError(
            "chennai.shp not found. Copy chennai.shp, .shx, .dbf and .prj "
            f"into {os.getcwd()} or set SHP_PATH to its full path."
        )

for p in (GREEN_PATH, NIR_PATH):
    if not os.path.exists(p):
        raise FileNotFoundError(f"Cannot find {p} in {os.getcwd()}")

# =========================================================
# 1. READ DATASETS
# =========================================================
chennai = gpd.read_file(SHP_PATH)

with rasterio.open(GREEN_PATH) as src_g:
    green = src_g.read(1).astype("float32")
    transform = src_g.transform
    raster_crs = src_g.crs
    out_shape = green.shape
    nodata_g = src_g.nodata

with rasterio.open(NIR_PATH) as src_n:
    nir = src_n.read(1).astype("float32")
    nodata_n = src_n.nodata

# Match boundary CRS to raster CRS
if chennai.crs != raster_crs:
    chennai = chennai.to_crs(raster_crs)

# =========================================================
# 2. NDWI = (Green - NIR) / (Green + NIR)
# =========================================================
green[green == -9999] = np.nan
nir[nir == -9999] = np.nan
if nodata_g is not None:
    green[green == nodata_g] = np.nan
if nodata_n is not None:
    nir[nir == nodata_n] = np.nan

with np.errstate(divide="ignore", invalid="ignore"):
    ndwi = (green - nir) / (green + nir)

# =========================================================
# 3. MASK TO CHENNAI BOUNDARY
# =========================================================
chennai_mask = geometry_mask(
    chennai.geometry,
    transform=transform,
    invert=True,           # True inside the boundary
    out_shape=out_shape,
)

ndwi[~chennai_mask] = np.nan

# =========================================================
# 4. PLOT NDWI
# =========================================================
plt.figure(figsize=(8, 7))

cmap = plt.cm.Blues.copy()
cmap.set_bad("white")      # NaN / outside area -> white

plt.imshow(ndwi, cmap=cmap, vmin=-1, vmax=1)
plt.colorbar(label="NDWI")
plt.title("Sentinel-2 NDWI - Chennai")
plt.axis("off")
plt.show()
```

```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import rasterio
from rasterio.features import geometry_mask
from rasterio.warp import reproject, Resampling

# =========================================================
# NDMI = Normalized Difference Moisture Index
#   NDMI = (NIR - SWIR1) / (NIR + SWIR1)
#   Sentinel-2: NIR = B8, SWIR1 = B11
# =========================================================

# =========================================================
# 0. FILE PATHS
# =========================================================
NIR_PATH   = r"B8.tif"     # NIR  (10 m)
SWIR_PATH  = r"B11.tif"    # SWIR1 (20 m)
SHP_PATH   = r"chennai.shp"
OUT_PATH   = r"Chennai_NDMI.tif"

# Find the shapefile if it is not in this folder
if not os.path.exists(SHP_PATH):
    search_root = os.path.dirname(os.getcwd())
    found = glob.glob(os.path.join(search_root, "**", "*hennai*.shp"), recursive=True)
    if found:
        SHP_PATH = found[0]
        print("Found shapefile at:", SHP_PATH)
    else:
        raise FileNotFoundError(
            "chennai.shp not found. Copy chennai.shp, .shx, .dbf and .prj "
            f"into {os.getcwd()} or set SHP_PATH to its full path."
        )

# =========================================================
# 1. READ BANDS
# =========================================================
with rasterio.open(NIR_PATH) as src_n:
    nir = src_n.read(1).astype("float32")
    profile = src_n.profile.copy()
    transform = src_n.transform
    raster_crs = src_n.crs
    out_shape = nir.shape
    nodata_n = src_n.nodata

# B11 is 20 m and B8 is 10 m -> resample B11 onto the B8 grid
with rasterio.open(SWIR_PATH) as src_s:
    nodata_s = src_s.nodata
    swir = np.full(out_shape, np.nan, dtype="float32")
    reproject(
        source=rasterio.band(src_s, 1),
        destination=swir,
        src_transform=src_s.transform,
        src_crs=src_s.crs,
        dst_transform=transform,
        dst_crs=raster_crs,
        resampling=Resampling.bilinear,
    )

# =========================================================
# 2. CLEAN NODATA AND CALCULATE NDMI
# =========================================================
nir[nir == -9999] = np.nan
swir[swir == -9999] = np.nan
if nodata_n is not None:
    nir[nir == nodata_n] = np.nan
if nodata_s is not None:
    swir[swir == nodata_s] = np.nan

with np.errstate(divide="ignore", invalid="ignore"):
    ndmi = (nir - swir) / (nir + swir)

# =========================================================
# 3. MASK TO CHENNAI BOUNDARY
# =========================================================
chennai = gpd.read_file(SHP_PATH)
if chennai.crs != raster_crs:
    chennai = chennai.to_crs(raster_crs)

chennai_mask = geometry_mask(
    chennai.geometry,
    transform=transform,
    invert=True,            # True inside the boundary
    out_shape=out_shape,
)
ndmi[~chennai_mask] = np.nan

# =========================================================
# 4. SAVE NDMI RASTER
# =========================================================
profile.update(dtype="float32", count=1, nodata=-99999)
ndmi_out = np.where(np.isnan(ndmi), -99999, ndmi).astype("float32")

with rasterio.open(OUT_PATH, "w", **profile) as dst:
    dst.write(ndmi_out, 1)
print("Saved:", OUT_PATH)

# =========================================================
# 5. PLOT NDMI
# =========================================================
plt.figure(figsize=(8, 7))

cmap = plt.cm.RdYlBu.copy()     # brown/red = dry, blue = moist
cmap.set_bad("white")           # outside Chennai -> white

plt.imshow(ndmi, cmap=cmap, vmin=-1, vmax=1)
plt.colorbar(label="NDMI")
plt.title("Sentinel-2 NDMI - Chennai")
plt.axis("off")
plt.show()
```

```python
import os, glob
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import rasterio
from rasterio.features import geometry_mask
from rasterio.warp import reproject, Resampling

# NDMI = (NIR - SWIR1) / (NIR + SWIR1)   ->  B8 and B11 (Sentinel-2)
NIR_PATH  = r"B8.tif"
SWIR_PATH = r"B11.tif"
SHP_PATH  = r"chennai.shp"
OUT_PATH  = r"Chennai_NDMI.tif"

# Find the shapefile if it is not in this folder
if not os.path.exists(SHP_PATH):
    root = os.path.dirname(os.getcwd())
    found = glob.glob(os.path.join(root, "**", "*hennai*.shp"), recursive=True)
    if not found:
        raise FileNotFoundError("chennai.shp not found. Copy it (with .shx, .dbf, .prj) into " + os.getcwd())
    SHP_PATH = found[0]

# 1. Read bands (B11 is 20 m, so resample it onto the B8 10 m grid)
with rasterio.open(NIR_PATH) as src_n:
    nir = src_n.read(1).astype("float32")
    profile = src_n.profile.copy()
    transform, crs, shape = src_n.transform, src_n.crs, nir.shape
    nodata_n = src_n.nodata

with rasterio.open(SWIR_PATH) as src_s:
    nodata_s = src_s.nodata
    swir = np.full(shape, np.nan, dtype="float32")
    reproject(
        source=rasterio.band(src_s, 1),
        destination=swir,
        src_transform=src_s.transform, src_crs=src_s.crs,
        dst_transform=transform, dst_crs=crs,
        resampling=Resampling.bilinear,
    )

# 2. Clean nodata and calculate NDMI
nir[nir == -9999] = np.nan
swir[swir == -9999] = np.nan
if nodata_n is not None: nir[nir == nodata_n] = np.nan
if nodata_s is not None: swir[swir == nodata_s] = np.nan

with np.errstate(divide="ignore", invalid="ignore"):
    ndmi = (nir - swir) / (nir + swir)

# 3. Mask to Chennai boundary
chennai = gpd.read_file(SHP_PATH)
if chennai.crs != crs:
    chennai = chennai.to_crs(crs)

mask = geometry_mask(chennai.geometry, transform=transform, invert=True, out_shape=shape)
ndmi[~mask] = np.nan

# 4. Save raster
profile.update(dtype="float32", count=1, nodata=-99999)
with rasterio.open(OUT_PATH, "w", **profile) as dst:
    dst.write(np.where(np.isnan(ndmi), -99999, ndmi).astype("float32"), 1)
print("Saved:", OUT_PATH)

# 5. Plot
cmap = plt.cm.RdYlBu.copy()
cmap.set_bad("white")
plt.figure(figsize=(8, 7))
plt.imshow(ndmi, cmap=cmap, vmin=-1, vmax=1)
plt.colorbar(label="NDMI")
plt.title("Sentinel-2 NDMI - Chennai")
plt.axis("off")
plt.show()
```
