```python
import matplotlib.pyplot as plt


df.groupby("region")["temperature"].mean().plot(kind="bar")

plt.xlabel("Region")
plt.ylabel("Average Temperature")
plt.title("Average Temperature by Region")
plt.show()
```

```python
import matplotlib.pyplot as plt


df.groupby("region")["rainfall"].mean().plot(kind="bar")

plt.xlabel("Region")
plt.ylabel("Average Rainfall")
plt.title("Average Rainfall by Region")
plt.show()
```

```python
import matplotlib.pyplot as plt


df.groupby("city")["rainfall"].mean().plot(kind="bar")

plt.xlabel("city")
plt.ylabel("Average Rainfall")
plt.title("Average Rainfall by City")
plt.show()
```

```python
import matplotlib.pyplot as plt


df.groupby("city")["rainfall"].mean().plot(kind="pie")

plt.xlabel("city")
plt.ylabel("Average Rainfall")
plt.title("Average Rainfall by City")
plt.show()
```

```python
import matplotlib.pyplot as plt


df.groupby("city")["ndvi"].mean().plot(kind="pie")

plt.xlabel("City")
plt.ylabel("Average NDVI")
plt.title("Average NDVI by City")
plt.show()
```
