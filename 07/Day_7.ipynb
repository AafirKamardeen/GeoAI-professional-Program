```python
import pandas as pd 
import matplotlib.pyplot as plt

df=pd.read_csv("Day7.csv")
```

```python
station_temp=df.groupby("station_name")["temperature"].mean()
```

```python
station_temp.plot(kind="bar")

plt.xlabel("Station")
plt.ylabel("Average Temperature (C)")
plt.title("Average Temerature by Station")

plt.tight_layout()
plt.show()
```

```python
station_temp.plot(
    kind="bar",
    color="red"
)

plt.xlabel("Station")
plt.ylabel("Average Temperature (C)")
plt.title("Average Temerature by Station")

plt.tight_layout()
plt.show()
```

```python
temp_range = df.groupby("station_name")["temperature"].agg(
    Min_Temperature="min",
    Max_Temperature="max"
)

temp_range
```

```python
temp_range.plot(kind="bar",
figsize=(10,5))

plt.xlabel("Station")
plt.ylabel("Average Temperature (C)")
plt.title("Minimum and maximum Temerature by Station")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.scatter(
    df["temperature"],
    df["ndvi"]
)

plt.xlabel("Temperature (C)")
plt.ylabel("NDVI")
plt.title(" Temerature  Vs NDVI")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.hist(
    df["temperature"],
    bins=8
)

plt.xlabel("Temperature (C)")
plt.ylabel("Usual")
plt.title(" Temerature  Vs Usaul")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.hist(
    df["temperature"],
    bins=8,
    color="green",
    edgecolor="black"
)

plt.xlabel("Temperature (C)")
plt.ylabel("Usual")
plt.title(" Temerature  Vs Usaul")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Python
import matplotlib.pyplot as plt
 
plt.scatter(
df["temperature"],
df["ndvi"]
)
 
plt.xlabel("Temperature (C)")
plt.ylabel("NDVI")
plt.title("Temperature vs NDVI")
 
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.boxplot(
    column="temperature",
    by="station_name"
)

plt.xlabel("Temperature (C)")
plt.ylabel("Station")
plt.title(" Temerature distribution by Station")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
