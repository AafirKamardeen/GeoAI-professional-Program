```python
import geopandas as gpd

# 1. Load the river polygon
rivers_poly = gpd.read_file("jamuna_polygon.geojson")

# 2. Check the current CRS (likely EPSG:4326, i.e. lat/lon degrees)
print("Original CRS:", rivers_poly.crs)

# 3. Reproject to a metric CRS (UTM zone picked automatically for the area)
utm_crs = rivers_poly.estimate_utm_crs()
rivers_poly_projected = rivers_poly.to_crs(utm_crs)

# 4. Check the result
print("Projected CRS:", rivers_poly_projected.crs)
rivers_poly_projected.total_bounds
```

```python
import geopandas as gpd

# 1. Load the river polygon
rivers_poly = gpd.read_file("jamuna_polygon.geojson")

# 2. Reproject to a metric CRS (needed for correct area in m²)
rivers_poly_projected = rivers_poly.to_crs(rivers_poly.estimate_utm_crs())

# 3. Area in square meters
rivers_poly_projected["Area_m2"] = rivers_poly_projected.geometry.area

# 4. Area in square kilometers
rivers_poly_projected["Area_km2"] = rivers_poly_projected["Area_m2"] / 1_000_000

# 5. Show selected columns
rivers_poly_projected[["rivname", "Area_m2", "Area_km2"]]
```

```python
import geopandas as gpd

# 1. Load the river line
rivers_line = gpd.read_file("jamuna.geojson")

# 2. Reproject to a metric CRS (so length is in meters)
rivers_line_projected = rivers_line.to_crs(rivers_line.estimate_utm_crs())

# 3. Calculate length in meters and store it in a new column
rivers_line_projected["length_m"] = rivers_line_projected.geometry.length

# 4. View the result
rivers_line_projected[["length_m"]]
```

```python
# TASK 2
# Convert Length_m into kilometers
# Create a new column called "length_km"

rivers_line_projected["length_km"] = rivers_line_projected["length_m"] / 1000

# View the result
rivers_line_projected[["length_m", "length_km"]]
```
